```python
import json
from google.colab import files
import os

# פונקציה לחלוקת הטקסט לנושאים ושמירתם במבנה נתונים
def parse_topics_to_array(text):
    # חלוקה לנושאים לפי הפרדה בין פסקאות
    topics = [topic.strip() for topic in text.split("\n\n") if topic.strip()]
    topics_data = []

    for topic in topics:
        # חלוקה לכותרת ותוכן
        lines = topic.split("\n", 1)
        if len(lines) > 1:
            title = lines[0].strip()
            content = lines[1].strip()
        else:
            title = "Untitled"
            content = lines[0].strip()

        # יצירת מבנה נתונים עבור הנושא
        topics_data.append({
            "topic": title,
            "content": content
        })

    return topics_data

# פונקציה לשמירת כל נושא בקובץ נפרד עם הצגת גודלו
def save_topics_to_files(topics_data):
    for topic in topics_data:
        topic_title = topic["topic"]
        topic_content = topic["content"]

        # יצירת שם קובץ ייחודי לכל נושא
        file_name = f"{topic_title.replace(' ', '_')}.txt"

        # שמירת התוכן של כל נושא בקובץ
        with open(file_name, "w", encoding="utf-8") as f:
            f.write(f"Topic: {topic_title}\n")
            f.write(f"Content: {topic_content}\n")

        # חישוב גודל הקובץ לאחר השמירה
        file_size = os.path.getsize(file_name) / 1024  # גודל בקילובייטים
        print(f"נושא '{topic_title}' נשמר בקובץ {file_name} (גודל: {file_size:.2f} KB)")

# העלאת קובץ
uploaded = files.upload()

# קריאת הקובץ
file_path = list(uploaded.keys())[0]
with open(file_path, 'r', encoding='utf-8') as f:
    text = f.read()

# עיבוד הטקסט למבנה נתונים
topics_data = parse_topics_to_array(text)

# שמירת הנושאים בקבצים נפרדים עם הצגת הגודל
save_topics_to_files(topics_data)
```

```python
import json
import nltk
import stanza

# הורדת הנתונים של Stanza אם לא קיימים
stanza.download('en')
nlp = stanza.Pipeline('en')

# פונקציה לניתוח SRL בעזרת Stanza
def analyze_srl(content):
    doc = nlp(content)
    results = []

    for sentence in doc.sentences:
        srl_result = {
            "sentence": " ".join([word.text for word in sentence.words]),
            "predicates": [word.text for word in sentence.words if "VB" in word.xpos],  # זיהוי פעלים
            "dependencies": [(word.text, word.head, word.deprel) for word in sentence.words]
        }
        results.append(srl_result)

    return results

# פונקציה ליצירת שאלות מה-SRL
def generate_questions_with_context(content, topic_context):
    results = analyze_srl(content)
    questions = []

    for result in results:
        sentence = result['sentence']
        predicates = result['predicates']

        for predicate in predicates:
            questions.append(f"What is the role of '{predicate}' in {topic_context}?")
            questions.append(f"How does '{predicate}' affect {topic_context}?")
            questions.append(f"Why is '{predicate}' important for {topic_context}?")

    return questions

# פונקציה לשמירת השאלות לקובץ טקסט
def save_questions_to_file(topic_title, questions):
    file_name = f"{topic_title.replace(' ', '_')}_questions.txt"
    with open(file_name, "w", encoding="utf-8") as file:
        for i, question in enumerate(questions, 1):
            file.write(f"{i}. {question}\n")
    print(f"Saved questions for '{topic_title}' to {file_name}")

generate_questions={}

for topic in topics_data:
    topic_title = topic["topic"]
    content = topic["content"]

    questions = generate_questions_with_context(content, topic_title)
    generate_questions[topic_title]=questions
```

```python
import re
import json
import torch
from sklearn.feature_extraction.text import TfidfVectorizer

# פונקציה לניקוי משפטים
def clean_sentence(sentence):
    return re.sub(r"<.*?>", "", sentence).strip()

# פונקציה למציאת המשפט המתאים ביותר עם TF-IDF
def find_best_match_tfidf(question, sentences):
    vectorizer = TfidfVectorizer()
    tfidf_matrix = vectorizer.fit_transform([question] + sentences)

    tfidf_tensor = torch.tensor(tfidf_matrix.toarray(), dtype=torch.float32)

    similarities = torch.nn.functional.cosine_similarity(
        tfidf_tensor[0:1], tfidf_tensor[1:], dim=1
    )

    top_indices = torch.argsort(similarities, descending=True)[:3]
    return [sentences[i] for i in top_indices]
# פונקציה למציאת אינדקסים של תשובה בתוך ההקשר
def find_answer_indices(context, answer):
    start_idx = context.find(answer)
    if start_idx == -1:
        return None, None
    end_idx = start_idx + len(answer) - 1
    return start_idx, end_idx

def process_topics_and_questions(topics_data, generated_questions, window_size=7):
    all_results = []
    used_answers = set()  # למניעת חזרות
    max_start = float('-inf')
    max_end = float('-inf')

    for topic in topics_data:
        topic_title = topic["topic"]
        content = topic["content"]

        sentences = [clean_sentence(sentence) for sentence in content.split(". ")]
        questions = generated_questions.get(topic_title, [])

        for question in questions:
            best_matches = find_best_match_tfidf(question, sentences)

            for best_match in best_matches:
                answer_text = best_match.strip()
                start_position, end_position = find_answer_indices(content, answer_text)

                # אם התשובה כבר קיימת, נמשיך למשפט הבא
                if answer_text in used_answers:
                    continue

                if start_position is not None and end_position is not None:
                    # חלוקת הטקסט למילים
                    content_tokens = content.split()
                    answer_tokens = answer_text.split()

                    # חישוב האינדקסים של מילת ההתחלה והסוף בקונטקסט
                    start_word_idx = max(0, start_position - window_size)
                    end_word_idx = min(len(content_tokens), end_position + len(answer_tokens) + window_size)

                    # שמירה רק על החלון סביב התשובה
                    reduced_context_tokens = content_tokens[start_word_idx:end_word_idx]
                    reduced_context = " ".join(reduced_context_tokens)

                    # אם ההקשר מצומצם ריק או קצר מדי, ננסה להגדיל את החלון
                    retries = 2  # מספר ניסיונות להפקת הקשר מדויק יותר
                    while (not reduced_context.strip() or len(reduced_context_tokens) < len(answer_tokens) + 5) and retries > 0:
                        window_size += 3  # הגדל את החלון
                        start_word_idx = max(0, start_position - window_size)
                        end_word_idx = min(len(content_tokens), end_position + len(answer_tokens) + window_size)
                        reduced_context_tokens = content_tokens[start_word_idx:end_word_idx]
                        reduced_context = " ".join(reduced_context_tokens)
                        retries -= 1

                    # אם ההקשר עדיין ריק לאחר ניסיונות, דלג על הרשומה
                    if not reduced_context.strip():
                        continue

                    # התאמת האינדקסים החדשים לתוך ההקשר המצומצם
                    new_start_position = start_position - start_word_idx
                    new_end_position = new_start_position + len(answer_text) - 1

                    # עדכון ה-max_start ו-max_end


                    if new_end_position > 1800:
                        continue


                    max_start = max(max_start, new_start_position)
                    max_end = max(max_end, new_end_position)
                    question_tokens = question.split()

                    all_results.append({
                        "question": question,
                        "context": reduced_context,
                        "question_tokens": question_tokens,
                        "context_tokens": reduced_context_tokens,
                        "start_position": new_start_position,
                        "end_position": new_end_position,
                    })

                    # הוספת התשובה לסט למניעת חזרות
                    used_answers.add(answer_text)

    print(f"Max start position: {max_start}")
    print(f"Max end position: {max_end}")

    return all_results

# הפקת הנתונים ושמירה בקובץ
processed_data = process_topics_and_questions(topics_data, generate_questions, window_size=7)
print(f"Number of answers generated: {len(processed_data)}")
output_file = "bidaf_ready_data.json"
with open(output_file, "w", encoding="utf-8") as f:
    json.dump(processed_data, f, ensure_ascii=False, indent=4)

print(f"Data saved in {output_file}")
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F


class CharacterEmbeddingLayer(nn.Module):
    def __init__(self, char_vocab_size, char_embedding_dim, char_cnn_kernel_size, char_output_dim, max_word_length):
        super(CharacterEmbeddingLayer, self).__init__()
        self.char_embedding = nn.Embedding(char_vocab_size, char_embedding_dim, padding_idx=0)

        kernel_height = min(char_cnn_kernel_size, max_word_length)
        print(f"Kernel Height: {kernel_height}, Max Word Length: {max_word_length}, Char Embedding Dim: {char_embedding_dim}")

        self.char_cnn = nn.Conv2d(
          in_channels=1,
          out_channels=256,
          kernel_size=(kernel_height, char_embedding_dim),
          padding=(0, char_embedding_dim // 2)
      )


    def forward(self, x):
      if len(x.size()) == 2:
          x = x.unsqueeze(0).unsqueeze(0)


      batch_size, seq_length, word_length = x.size()
      x = x.view(-1, word_length)

      # Validate indices before embedding
      x = torch.clamp(x, 0, self.char_embedding.num_embeddings - 1)

      x = self.char_embedding(x)  # Embedding

      x = x.unsqueeze(1)  # Add channel dimension for CNN
      x = self.char_cnn(x)  # Apply CNN

      x = F.relu(x)

      # Apply max pooling over the correct dimension
      x = F.max_pool2d(x, kernel_size=(x.size(2), x.size(3)))

      x = x.view(batch_size, seq_length, -1)  # Restore original dimensions
      return x






class WordEmbeddingLayer(nn.Module):
    def __init__(self, word_vocab_size, word_embedding_dim, pretrained_embeddings=None):
        super(WordEmbeddingLayer, self).__init__()
        self.word_embedding = nn.Embedding(word_vocab_size, word_embedding_dim, padding_idx=0)

        if pretrained_embeddings is not None:
            self.word_embedding.weight.data.copy_(pretrained_embeddings)
            self.word_embedding.weight.requires_grad = False

    def forward(self, x):
        # Check for out-of-range indices
        assert x.min().item() >= 0, f"Min index in x is {x.min().item()}, which is negative"
        assert x.max().item() < self.word_embedding.num_embeddings, f"Max index in x is {x.max().item()}, which exceeds vocab size of {self.word_embedding.num_embeddings}"

        return self.word_embedding(x)


class BiLSTMLayer(nn.Module):
    def __init__(self, input_dim, hidden_dim, num_layers=1, dropout=0.0):
        super(BiLSTMLayer, self).__init__()
        self.bilstm = nn.LSTM(
            input_dim, hidden_dim, num_layers=num_layers,
            bidirectional=True, batch_first=True,
            dropout=dropout if num_layers > 1 else 0.0
        )

    def forward(self, x, lengths):
        packed_x = nn.utils.rnn.pack_padded_sequence(
            x, lengths, batch_first=True, enforce_sorted=False
        )
        packed_output, _ = self.bilstm(packed_x)
        output, _ = nn.utils.rnn.pad_packed_sequence(packed_output, batch_first=True)
        return output


class BiDAFEncoder(nn.Module):
    def __init__(self, char_vocab_size, char_embedding_dim, char_cnn_kernel_size,
                 char_output_dim, word_vocab_size, word_embedding_dim, hidden_dim,
                 max_word_length, pretrained_word_embeddings=None):
        super(BiDAFEncoder, self).__init__()
        self.char_embed = CharacterEmbeddingLayer(
            char_vocab_size, char_embedding_dim, char_cnn_kernel_size, char_output_dim, max_word_length
        )
        self.word_embed = WordEmbeddingLayer(word_vocab_size, word_embedding_dim, pretrained_word_embeddings)

        # הגדרת input_dim בהתבסס על החיבור של שתי השכבות
        input_dim = char_output_dim + word_embedding_dim
        self.bilstm = BiLSTMLayer(input_dim, hidden_dim)


    def forward(self, context, context_lengths, query, query_lengths, char_context, char_query):
        # Character Embeddings
        char_context_emb = self.char_embed(char_context)  # (batch_size, context_length, char_output_dim)
        char_query_emb = self.char_embed(char_query)  # (batch_size, 1, char_output_dim)

        # Align dimensions of char_query_emb to match word_query_emb
        char_query_emb = char_query_emb.expand(-1, query.size(1), -1)  # (batch_size, query_length, char_output_dim)

        # Word Embeddings
        word_context_emb = self.word_embed(context)  # (batch_size, context_length, word_embedding_dim)
        word_query_emb = self.word_embed(query)  # (batch_size, query_length, word_embedding_dim)

        # Concatenate character and word embeddings
        context_emb = torch.cat([char_context_emb, word_context_emb], dim=2)  # (batch_size, context_length, input_dim)
        query_emb = torch.cat([char_query_emb, word_query_emb], dim=2)  # (batch_size, query_length, input_dim)

        # Apply BiLSTM
        context_output = self.bilstm(context_emb, context_lengths)
        query_output = self.bilstm(query_emb, query_lengths)

        return context_output, query_output
```

```python
class BiDAFAttentionLayer(nn.Module):
    def __init__(self, hidden_dim):
        super(BiDAFAttentionLayer, self).__init__()
        self.hidden_dim = hidden_dim

        # Adjusted the Linear layer input size
        self.W = nn.Linear(hidden_dim * 3, 1, bias=False)  # hidden_dim * 8 is expected after concatenation

    def forward(self, context, query):
        batch_size, context_length, _ = context.size()
        _, query_length, _ = query.size()

        # Expand context and query for broadcasting
        context_expanded = context.unsqueeze(2).expand(-1, -1, query_length, -1)  # (batch_size, context_length, query_length, hidden_dim * 2)
        query_expanded = query.unsqueeze(1).expand(-1, context_length, -1, -1)  # (batch_size, context_length, query_length, hidden_dim * 2)

        # Calculate elementwise product for similarity computation
        elementwise_product = context_expanded * query_expanded  # (batch_size, context_length, query_length, hidden_dim * 2)

        # Concatenate context, query, and elementwise product (resulting in hidden_dim * 8)
        similarity_input = torch.cat([context_expanded, query_expanded, elementwise_product], dim=-1)  # (batch_size, context_length, query_length, hidden_dim * 8)


        # Apply Linear layer to similarity_input
        similarity_matrix = self.W(similarity_input).squeeze(-1)  # (batch_size, context_length, query_length)

        # Compute context-to-query attention (c2q_attention)
        c2q_attention = F.softmax(similarity_matrix, dim=-1)  # (batch_size, context_length, query_length)
        attended_query = torch.bmm(c2q_attention, query)  # (batch_size, context_length, hidden_dim * 2)

        # Compute query-to-context attention (q2c_attention)
        q2c_attention = F.softmax(similarity_matrix.max(dim=2).values, dim=-1)  # (batch_size, context_length)
        attended_context = torch.bmm(q2c_attention.unsqueeze(1), context).squeeze(1)  # (batch_size, hidden_dim * 2)

        attended_context = attended_context.unsqueeze(1).expand(-1, context_length, -1)  # (batch_size, context_length, hidden_dim * 2)

        # Combine context, attended query, and attended context
        G = torch.cat([context, attended_query, context * attended_query, attended_context * context], dim=-1)  # (batch_size, context_length, hidden_dim * 8)


        return G
```

```python
class ModelingLayer(nn.Module):
    def __init__(self, input_dim, hidden_dim, num_layers=2, dropout=0.2):
        super(ModelingLayer, self).__init__()
        self.bilstm = nn.LSTM(
            input_dim, hidden_dim, num_layers=num_layers,
            bidirectional=True, batch_first=True, dropout=dropout
        )

    def forward(self, x):
        output, _ = self.bilstm(x)
        return output
```

```python
class OutputLayer(nn.Module):
    def __init__(self, hidden_dim):
        super(OutputLayer, self).__init__()
        self.start_layer = nn.Linear(750, 1)  # התאמה למימדים שלך
        self.end_layer = nn.Linear(750, 1)
        self.end_bilstm = nn.LSTM(hidden_dim * 2, hidden_dim, num_layers=1, bidirectional=True, batch_first=True)

    def forward(self, attention_output, modeling_output):
        # Concatenate attention_output and modeling_output
        combined = torch.cat([attention_output, modeling_output], dim=-1)

        # Flatten for Linear layer
        combined_flattened = combined.view(-1, combined.size(-1))

        # Compute start_logits
        start_logits = self.start_layer(combined_flattened).squeeze(-1)

        # Process end_logits through BiLSTM
        end_modeling, _ = self.end_bilstm(modeling_output)

        # Concatenate attention_output with end_modeling
        end_combined = torch.cat([attention_output, end_modeling], dim=-1)

        # Flatten for end_combined
        end_combined_flattened = end_combined.view(-1, end_combined.size(-1))

        # Compute end_logits
        end_logits = self.end_layer(end_combined_flattened).squeeze(-1)

        return (
            start_logits.view(attention_output.size(0), -1),
            end_logits.view(attention_output.size(0), -1)
        )
```

```python
def bidaf_loss(start_logits, end_logits, start_positions, end_positions):
    start_loss = F.cross_entropy(start_logits, start_positions)
    end_loss = F.cross_entropy(end_logits, end_positions)
    return start_loss + end_loss
```

```python
class BiDAFModel(nn.Module):
    def __init__(self, char_vocab_size, char_embedding_dim, char_cnn_kernel_size, char_output_dim,
                 word_vocab_size, word_embedding_dim, hidden_dim, max_word_length,  dropout_rate=0.2, pretrained_word_embeddings=None):
        super(BiDAFModel, self).__init__()

        # Encoder layers
        self.encoder = BiDAFEncoder(
            char_vocab_size, char_embedding_dim, char_cnn_kernel_size, char_output_dim,
            word_vocab_size, word_embedding_dim, hidden_dim, max_word_length, pretrained_word_embeddings
        )

        # Attention layer
        self.attention = BiDAFAttentionLayer(hidden_dim * 2)

        # Modeling layer
        self.modeling = ModelingLayer(hidden_dim * 8, hidden_dim)

        self.dropout = nn.Dropout(dropout_rate)
        # Output layer
        self.output = OutputLayer(hidden_dim)

    def forward(self, context, context_lengths, query, query_lengths, char_context, char_query):

        # Fix dimensions if needed
        if len(char_query.size()) == 2:
            char_query = char_query.unsqueeze(1)

        # Encode context and query using BiDAFEncoder
        context_output, query_output = self.encoder(
            context, context_lengths, query, query_lengths, char_context, char_query
        )

        # Apply attention
        attention_output = self.attention(context_output, query_output)

        # Apply modeling
        modeling_output = self.modeling(attention_output)

        modeling_output = self.dropout(modeling_output)

        # Combine attention and modeling outputs
        combined = torch.cat([attention_output, modeling_output], dim=-1)

        # Get start and end logits
        start_logits, end_logits = self.output(attention_output, modeling_output)

        return start_logits, end_logits
```

```python
import torch
from torch.cuda.amp import autocast, GradScaler
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from tqdm import tqdm
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt
import json
from collections import defaultdict
import os
from torch.optim.lr_scheduler import ReduceLROnPlateau
#os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"

device = torch.device("cpu")
print(f"Using device: {device}")

# ----------------------------------------
# Load train data
# ----------------------------------------
with open("train_data.json", "r") as f:
    train_data = json.load(f)

for item in train_data:
    context_tokens = item["context_tokens"]
    question_tokens = item["question_tokens"]
    item["start_position"] = min(item["start_position"], len(item["context_tokens"]) - 1)
    item["end_position"] = min(item["end_position"], len(item["context_tokens"]) - 1)

# ----------------------------------------
# Load test data
# ----------------------------------------
with open("test_data.json", "r") as f:
    test_data = json.load(f)

for item in test_data:
    context_tokens = item["context_tokens"]
    question_tokens = item["question_tokens"]
    item["start_position"] = min(item["start_position"], len(item["context_tokens"]) - 1)
    item["end_position"] = min(item["end_position"], len(item["context_tokens"]) - 1)

# ----------------------------------------
# Build Vocabulary
# ----------------------------------------
def build_vocab(data):
    vocab = defaultdict(lambda: 1)  # Default to [UNK] (Unknown Token)
    vocab["[PAD]"] = 0  # Padding token
    vocab["[UNK]"] = 1  # Unknown token (OOV)

    index = 2  # התחל מ-2 כי 0 שמור ל-[PAD] ו-1 שמור ל-[UNK]

    for item in data:
        for token in item["question_tokens"] + item["context_tokens"]:
            if token not in vocab:
                vocab[token] = index
                index += 1

    return vocab


def build_char_vocab(data):
    char_vocab = defaultdict(lambda: 1)  # Default to 1 for unknown chars
    char_vocab["[PAD]"] = 0  # Padding token
    index = 2
    for item in data:
        for token in item["question_tokens"] + item["context_tokens"]:
            for char in token:
                if char not in char_vocab:
                    char_vocab[char] = index
                    index += 1
    return char_vocab

word_vocab = build_vocab(train_data)
char_vocab = build_char_vocab(train_data)


# ----------------------------------------
# Early Stopping class
# ----------------------------------------

class EarlyStopping:
    def __init__(self, patience=2, delta=-0.5):
        self.patience = patience
        self.delta = delta
        self.best_loss = None
        self.counter = 0
        self.early_stop = False

    def __call__(self, val_loss):
        if self.best_loss is None:
            self.best_loss = val_loss
        elif val_loss > self.best_loss + self.delta:
            self.counter += 1
            print(f"EarlyStopping counter: {self.counter} out of {self.patience}")
            if self.counter >= self.patience:
                self.early_stop = True
        else:
            self.best_loss = val_loss
            self.counter = 0

# ----------------------------------------
# Dataset Definition
# ----------------------------------------
class QADataset(Dataset):
    def __init__(self, data, word_vocab, char_vocab, max_context_length=2000, max_question_length=500, max_word_length=10):
        self.data = data
        self.word_vocab = word_vocab
        self.char_vocab = char_vocab
        self.max_context_length = max_context_length
        self.max_question_length = max_question_length
        self.max_word_length = max_word_length

    def __len__(self):
        return len(self.data)

    def _token_to_char_indices(self, tokens):
        char_indices = [
            [self.char_vocab[char] for char in token[:self.max_word_length]]
            + [0] * (self.max_word_length - len(token))
            for token in tokens
        ]
        return char_indices

    def __getitem__(self, idx):
        item = self.data[idx]
        question_tokens = item["question_tokens"]
        context_tokens = item["context_tokens"][:9000]
        # Word-level input
        question = [self.word_vocab[token] for token in question_tokens[:self.max_question_length]]
        context = [self.word_vocab[token] for token in context_tokens[:self.max_context_length]]

        max_vocab_idx = len(word_vocab) - 1
        question = [min(word_vocab.get(token, 1), max_vocab_idx) for token in question_tokens[:500]]
        context = [min(word_vocab.get(token, 1), max_vocab_idx) for token in context_tokens[:2000]]




        question += [0] * (self.max_question_length - len(question))
        context += [0] * (self.max_context_length - len(context))

        # Character-level input
        char_question = self._token_to_char_indices(question_tokens[:self.max_question_length])
        char_context = self._token_to_char_indices(context_tokens[:self.max_context_length])

        char_question += [[0] * self.max_word_length] * (self.max_question_length - len(char_question))
        char_context += [[0] * self.max_word_length] * (self.max_context_length - len(char_context))

        return {
            "question": torch.tensor(question, dtype=torch.long),
            "context": torch.tensor(context, dtype=torch.long),
            "char_question": torch.tensor(char_question, dtype=torch.long),
            "char_context": torch.tensor(char_context, dtype=torch.long),
            "start_position": torch.tensor(item["start_position"], dtype=torch.long),
            "end_position": torch.tensor(item["end_position"], dtype=torch.long),
        }

# DataLoader for training and testing
train_dataset = QADataset(train_data, word_vocab, char_vocab)
train_loader = DataLoader(train_dataset, batch_size=16, shuffle=True)

test_dataset = QADataset(test_data, word_vocab, char_vocab)
test_loader = DataLoader(test_dataset, batch_size=16, shuffle=False)

# ----------------------------------------
# BiDAF Model Definition
# ----------------------------------------
model = BiDAFModel(
    char_vocab_size=len(char_vocab),
    char_embedding_dim=16,
    char_cnn_kernel_size=3,
    char_output_dim=256,
    word_vocab_size=len(word_vocab),
    word_embedding_dim=32,
    hidden_dim=75,
    max_word_length=15,
    dropout_rate=0.5
)

# ----------------------------------------
# Loss Function
# ----------------------------------------
def bidaf_loss(start_logits, end_logits, start_positions, end_positions):
    start_loss = F.cross_entropy(start_logits, start_positions)
    end_loss = F.cross_entropy(end_logits, end_positions)
    return start_loss + end_loss

# ----------------------------------------
# Compute F1 and Compute Exact Match
# ----------------------------------------

def compute_f1(true_start, true_end, pred_start, pred_end):
    true_set = set(range(true_start, true_end + 1))
    pred_set = set(range(pred_start, pred_end + 1))
    common = true_set & pred_set
    if len(common) == 0:
        return 0.0
    precision = len(common) / len(pred_set)
    recall = len(common) / len(true_set)
    f1 = 2 * (precision * recall) / (precision + recall)
    return f1

def compute_exact_match(true_start, true_end, pred_start, pred_end):
    return int(true_start == pred_start and true_end == pred_end)

# ----------------------------------------
# Evaluate Model
# ----------------------------------------

def evaluate_model(model, test_loader, device="cpu"):
    model.eval()
    total_loss, total_f1, total_em, total_examples = 0, 0, 0, 0
    all_start_preds = []
    all_start_true = []
    all_end_preds = []
    all_end_true = []

    with torch.no_grad():
        for batch in tqdm(test_loader):
            question = batch["question"].to(device)
            context = batch["context"].to(device)
            char_question = batch["char_question"].to(device)
            char_context = batch["char_context"].to(device)
            start_positions = batch["start_position"].to(device)
            end_positions = batch["end_position"].to(device)

            context_lengths = torch.tensor([min(context.size(1), 4000) for _ in range(context.size(0))], dtype=torch.long).cpu()
            query_lengths = torch.tensor([min(question.size(1), 500) for _ in range(question.size(0))], dtype=torch.long).cpu()

            start_logits, end_logits = model(
                context=context,
                context_lengths=context_lengths,
                query=question,
                query_lengths=query_lengths,
                char_context=char_context,
                char_query=char_question
            )



            loss = bidaf_loss(start_logits, end_logits, start_positions, end_positions)
            total_loss += loss.item()

            pred_start = start_logits.argmax(dim=1)
            pred_end = end_logits.argmax(dim=1)

            for i in range(start_positions.size(0)):
                true_start = start_positions[i].item()
                true_end = end_positions[i].item()
                pred_start_idx = pred_start[i].item()
                pred_end_idx = pred_end[i].item()

                all_start_preds.append(pred_start_idx)
                all_start_true.append(true_start)
                all_end_preds.append(pred_end_idx)
                all_end_true.append(true_end)

                total_f1 += compute_f1(true_start, true_end, pred_start_idx, pred_end_idx)
                total_em += compute_exact_match(true_start, true_end, pred_start_idx, pred_end_idx)

            total_examples += start_positions.size(0)

    avg_loss = total_loss / len(test_loader)
    avg_f1 = total_f1 / total_examples
    avg_em = total_em / total_examples

    print(f"Test Results - Loss: {avg_loss:.4f}, F1: {avg_f1:.4f}, Exact Match: {avg_em:.4f}")
    test_losses.append(avg_loss)
    test_f1_scores.append(avg_f1)
    test_em_scores.append(avg_em)



# ----------------------------------------
# Training Loop and Evaluate after each epoch
# ----------------------------------------

def train_model(model, train_loader, test_loader, optimizer, epochs=10, device="cpu"):
    model.to(device)
    scaler = GradScaler()
   # early_stopping = EarlyStopping(patience=2, delta=-0.5)

    for epoch in range(epochs):
        model.train()
        total_loss = 0
        total_em = 0
        total_f1 = 0
        total_examples = 0

        for batch in tqdm(train_loader):
            question = batch["question"].to(device)
            context = batch["context"].to(device)
            char_question = batch["char_question"].to(device)
            char_context = batch["char_context"].to(device)
            start_positions = batch["start_position"].to(device)
            end_positions = batch["end_position"].to(device)

            context_lengths = torch.tensor([min(context.size(1), 4000) for _ in range(context.size(0))], dtype=torch.long).cpu()
            query_lengths = torch.tensor([min(question.size(1), 500) for _ in range(question.size(0))], dtype=torch.long).cpu()

            optimizer.zero_grad()
            with autocast():
                start_logits, end_logits = model(
                    context=context,
                    context_lengths=context_lengths,
                    query=question,
                    query_lengths=query_lengths,
                    char_context=char_context,
                    char_query=char_question
                )

                loss = bidaf_loss(start_logits, end_logits, start_positions, end_positions)

            scaler.scale(loss).backward()
            scaler.step(optimizer)
            scaler.update()

            total_loss += loss.item()

            batch_size = start_positions.size(0)
            total_examples += batch_size

            pred_start = start_logits.argmax(dim=1)
            pred_end = end_logits.argmax(dim=1)

            for i in range(batch_size):
                true_start = start_positions[i].item()
                true_end = end_positions[i].item()
                pred_start_idx = pred_start[i].item()
                pred_end_idx = pred_end[i].item()

                total_f1 += compute_f1(true_start, true_end, pred_start_idx, pred_end_idx)
                total_em += compute_exact_match(true_start, true_end, pred_start_idx, pred_end_idx)

        avg_loss = total_loss / len(train_loader)
        avg_f1 = total_f1 / total_examples
        avg_em = total_em / total_examples


        print(f"Epoch {epoch + 1}/{epochs}, Loss: {avg_loss:.4f}, F1: {avg_f1:.4f}, Exact Match: {avg_em:.4f}")

        train_losses.append(avg_loss)
        train_f1_scores.append(avg_f1)
        train_em_scores.append(avg_em)

        # Evaluate the model after each epoch
        evaluate_model(model, test_loader, device)
        print(avg_loss)
        scheduler.step(avg_loss)


# ----------------------------------------
# Plotting the Results
# ----------------------------------------

def plot_results():
    epochs = range(1, len(train_losses) + 1)

    # Plotting the Loss
    plt.figure(figsize=(12, 6))
    plt.subplot(1, 3, 1)
    plt.plot(epochs, train_losses, label="Train Loss")
    plt.plot(epochs, test_losses, label="Test Loss")
    plt.xlabel('Epochs')
    plt.ylabel('Loss')
    plt.legend()

    # Plotting F1 Score
    plt.subplot(1, 3, 2)
    plt.plot(epochs, train_f1_scores, label="Train F1")
    plt.plot(epochs, test_f1_scores, label="Test F1")
    plt.xlabel('Epochs')
    plt.ylabel('F1 Score')
    plt.legend()

    # Plotting Exact Match Score
    plt.subplot(1, 3, 3)
    plt.plot(epochs, train_em_scores, label="Train Exact Match")
    plt.plot(epochs, test_em_scores, label="Test Exact Match")
    plt.xlabel('Epochs')
    plt.ylabel('Exact Match')
    plt.legend()

    plt.tight_layout()
    plt.show()


# ----------------------------------------
# Optimizer and Training
# ----------------------------------------

optimizer = torch.optim.Adam(model.parameters(), lr=0.001, weight_decay=0.01)

scheduler = ReduceLROnPlateau(optimizer, 'min', patience=2, verbose=True, factor=0.5)

train_losses = []
train_f1_scores = []
train_em_scores = []
test_losses = []
test_f1_scores = []
test_em_scores = []

train_model(model, train_loader, test_loader, optimizer, epochs=13)
torch.save(model.state_dict(), "bidaf_model.pth")
print("Model saved to bidaf_model.pth")

plot_results()
```
